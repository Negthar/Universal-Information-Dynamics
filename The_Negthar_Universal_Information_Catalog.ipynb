### Overview
This notebook documents a revolutionary approach to understanding the physical universe through the lens of **Information Physics**. According to the **1-5-6 Model**, matter is not a fundamental entity but an emergent property (Output) of high-frequency information processing.

### The Core Formula
The transition from Information to Mass is governed by the **Negthar Constant ($K_N$)**:

$$m = \frac{h \cdot \nu_{info} \cdot K_N \cdot N}{c^2}$$

Where:
* $m$: Physical Mass (kg)
* $h$: Planck Constant
* $\nu_{info}$: Intrinsic Information Frequency ($10^{11}$ Hz)
* $K_N$: Negthar Constant ($50,000$)
* $N$: Information Cycles (The complexity of the particle)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Universal Constants
PLANCK_H = 6.62607015e-34  # Planck constant (h)
LIGHT_C = 299792458        # Speed of light (c)

# Negthar Framework Constants
NU_INFO = 1e11             # Intrinsic Frequency (10^11 Hz)
KN = 50000                 # Negthar Constant (Scaling Factor)

print("Negthar Engine Initialized.")
print(f"KN Value: {KN}")
print(f"System Frequency: {NU_INFO} Hz")
```

### 1. Mapping Fundamental Particles to Information Cycles (N)
In this section, we convert the officially measured mass of elementary particles into their corresponding **Information Cycles (N)**. This represents the internal code complexity required by the universe to "render" each particle.

```python
# List of standard particles with their measured masses (kg)
particles = {
    "Particle": ["Electron", "Proton", "Neutron", "Muon", "Tau"],
    "Mass_kg": [9.1093837e-31, 1.6726219e-27, 1.6749275e-27, 1.8835316e-28, 3.16747e-27]
}

df = pd.DataFrame(particles)

# Function to calculate N (Information Cycles)
def calculate_n_cycles(mass):
    return (mass * LIGHT_C**2) / (PLANCK_H * NU_INFO * KN)

# Applying the 1-5-6 Model
df['N_Cycles'] = df['Mass_kg'].apply(calculate_n_cycles)
df['N_Integer'] = df['N_Cycles'].round().astype(int)
df['Processing_Noise'] = df['N_Cycles'] - df['N_Integer']

df
```

### 2. Visualization of Information Complexity
Since the difference in $N$ values between particles (e.g., Electron vs. Tau) is vast, we use a **Logarithmic Scale** to visualize the information complexity. This chart illustrates the relative "computational weight" the universe assigns to each particle type.

```python
import matplotlib.pyplot as plt

# Setting up the plot
plt.figure(figsize=(12, 7))
colors = ['#3498db', '#e74c3c', '#2ecc71', '#f1c40f', '#9b59b6']

# Creating the bar chart
bars = plt.bar(df['Particle'], df['N_Cycles'], color=colors, alpha=0.8)

# Applying Logarithmic Scale for better visibility of vast differences
plt.yscale('log')

# Adding labels and title
plt.xlabel('Fundamental Particles', fontsize=12)
plt.ylabel('Information Cycles (N) [Log Scale]', fontsize=12)
plt.title('Informational Mapping: Complexity of Particles (1-5-6 Model)', fontsize=14, pad=20)
plt.grid(True, which="both", ls="--", alpha=0.5)

# Adding the exact N values on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, f'{yval:.2e}',
             va='bottom', ha='center', fontsize=10, fontweight='bold')

plt.show()
```

```python
# --- SECTION 7: INITIAL OVERHEAD ANALYSIS ---
# Goal: Calculate the raw noise (epsilon) for each fundamental particle.

# 1. Calculate the total visible information (Integer N)
total_n_integer = df['N_Integer'].sum()

# 2. Calculate the total processing noise (Absolute epsilon)
total_epsilon = df['Processing_Noise'].abs().sum()

# 3. Calculate the local noise-to-mass ratio
local_overhead_ratio = (total_epsilon / (total_n_integer + total_epsilon)) * 100

print(f"Total Visible Cycles (N_integer): {total_n_integer}")
print(f"Total Processing Noise (epsilon): {total_epsilon:.4f}")
print(f"Local Noise Ratio: {local_overhead_ratio:.6f}%")
```
